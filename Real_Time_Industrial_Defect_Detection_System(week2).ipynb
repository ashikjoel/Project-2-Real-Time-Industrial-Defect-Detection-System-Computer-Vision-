#YOLOv8n (Nano) – Baseline Model (image size=640)

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data="/content/drive/MyDrive/Industrial_Defect_Detection/yolo_dataset/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    project="/content/drive/MyDrive/Industrial_Defect_Detection/runs",
    name="yolov8n_baseline"
)
```

#Precision   : 70.5% ,Recall      : 69.2% ,mAP50       : 74.7% ,mAP50-95    : 39.2%



```python
import os

results_path = "/content/drive/MyDrive/Industrial_Defect_Detection/runs/yolov8n_baseline"

for file in os.listdir(results_path):
    print(file)
```
